# Adjust for Holidays

In this section we test some calendar functionalities of `tensorquant` using the `TARGET` calendar:

- **Business day check**: `is_business_day(date)` returns whether a given date is a business day.
- **Holiday check**: `is_holiday(date)` tells you if the date is a holiday according to the chosen calendar.
- **End of month**: `is_end_of_month(date)` checks whether the date falls on the end of the month (with or without business-day adjustment).
- **Date adjustment**: `adjust(date, BusinessDayConvention)` shifts the date according to the chosen convention (`Following`, `ModifiedFollowing`, `Preceding`, etc.).

The example uses **Easter Monday 2024** to show how the calendar recognizes a holiday and how the different adjustment conventions lead to different dates.

```python

d = date(2024, 3, 31) # easter monday 2024
calendar_test = tq.TARGET()
```

```python
print('Is BusinessDay?: ' , calendar_test.is_business_day(d))
print('Is Holiday?: ' , calendar_test.is_holiday(d))
```

```python

print('Is EoM Unadjasted?: ' , calendar_test.is_end_of_month(d))
```

```python

print("Following: ", calendar_test.adjust(d, tq.BusinessDayConvention.Following))
print("ModFol: ", calendar_test.adjust(d, tq.BusinessDayConvention.ModifiedFollowing)) # equal to preceding since it's an EoM
print("Preceding: ", calendar_test.adjust(d, tq.BusinessDayConvention.Preceding))
```

# Advance Dates

In this section we use `calendar_test.advance(...)` to move a date into the future by a given number of periods (`n_period`) and time units (`TimeUnit`), applying a chosen **adjustment convention**.

In this specific case:

- **Starting date**: `start_date = date(2024, 1, 31)`.
- **Number of periods**: `n_period = 2`.
- **Time unit**: `TimeUnit.Months` (so 2 months).
- **Conventions tested**:
  - `ModifiedFollowing`: if the date falls at the end of the month and the shifted date is not a business day, it takes **the last business day of that month**.
  - `Following`: simply moves to the **next business day**, without preserving the end-of-month.

The examples clearly show the difference between keeping the "end-of-month" effect (ModifiedFollowing) and ignoring it (Following).

```python
start_date = date(2024, 1, 31)
n_period = 2
period_unit =  tq.TimeUnit.Months

advanced_date = calendar_test.advance(start_date,
                                        n_period, 
                                        period_unit,
                                        tq.BusinessDayConvention.ModifiedFollowing 
                                        )
```

```python
# 29-30-31 March are holidays
# EoM --> take last BD of the month
advanced_date
```

```python
advanced_date2 = calendar_test.advance(start_date,
                                        n_period, 
                                        period_unit,
                                        tq.BusinessDayConvention.Following
                                        )
```

```python
# Folowing --> don't care about EoM, just give me the next BD
advanced_date2
```
